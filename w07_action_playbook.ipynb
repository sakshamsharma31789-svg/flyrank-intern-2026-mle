### 1. Ranked actions + reason codes
We construct a prioritized queue of content actions based on model prediction scores, freshness decay, and historical engagement lifts. Each recommended action is accompanied by a transparent reason code to ensure human operators understand the underlying rationale.

```python
import pandas as pd

# Example code building a sample ranked action queue with reason codes
data = [
    {"content_id": "CNT-1042", "archetype": "deep_dive", "action": "feature_refresh", "score": 0.94, "reason_code": "HIGH_CONFIDENCE_FRESHNESS_DECAY"},
    {"content_id": "CNT-2081", "archetype": "quick_take", "action": "archive_or_redirect", "score": 0.88, "reason_code": "LOW_ENGAGEMENT_VELOCITY"},
    {"content_id": "CNT-3019", "archetype": "tutorial", "action": "promote_featured", "score": 0.82, "reason_code": "HIGH_CONVERSION_POTENTIAL"}
]

queue_df = pd.DataFrame(data)
print("Ranked Action Queue Preview:")
display(queue_df.head())
```

### 2. Intended use and limits
- **Intended Use:** Decision-support and operational prioritization for editorial teams. The system surfaces observed performance patterns to assist human reviewers in choosing content interventions.
- **Limits & Scope:** The model's validity is bounded by historical distribution bounds. It assumes stationary user intent and normal traffic conditions; it is not reliable for out-of-distribution events, sudden viral shifts, or sparse legacy items. All claims derived here are strictly directional and measured.

```python
# Validation check for scope boundaries and feature null-checks
def validate_operational_limits(df):
    assert not df.empty, "Queue cannot be empty."
    print(f"Validated {len(df)} records within operational scope limits.")

validate_operational_limits(queue_df)
```

### 3. Human review + the no-go list
- **Human Review Checklist:**
  1. Verify brand safety and tone compliance.
  2. Confirm factual accuracy of underlying content updates.
  3. Check for seasonal or temporal conflicts.
- **The No-Go List (Never Automate):**
  - Direct un-reviewed deletions or archiving of high-traffic flagship content.
  - Automated generation and publishing of sensitive financial, legal, or health claims.
  - Blind mass-redirection of historical URLs without link equity validation.

```python
# Code check enforcing human review flags
queue_df["requires_human_signoff"] = True
print("Human review gate enabled for all queued actions:", queue_df["requires_human_signoff"].all())
```

### 4. Monitoring / retrain triggers
The model and ranking logic will be considered stale or in need of retraining under the following conditions:
- **Distribution Drift:** Kolmogorov-Smirnov test p-value drops below 0.05 on core feature inputs week-over-week.
- **Performance Decay:** A sustained 15% drop in recommendation click-through or action completion rate over a 14-day rolling window.
- **Freshness Threshold:** Median content age in the top-ranked queue exceeds allowable temporal bounds.

```python
# Simulated monitoring threshold configuration check
monitoring_config = {
    "max_ks_drift_p_value": 0.05,
    "max_ctr_drop_pct": 15.0,
    "rolling_window_days": 14
}
print("Monitoring configuration active:", monitoring_config)
```

### 5. Exports for the paper
Exporting the finalized action queue to work/outputs/ (dynamically generated per run) and saving summary figures for direct inclusion into next week's research paper build.

```python
import os
import matplotlib.pyplot as plt

# Ensure output and figure directories exist
os.makedirs("work/outputs", exist_ok=True)
os.makedirs("work/figures", exist_ok=True)

# Export ranked queue CSV
output_path = "work/outputs/action_queue.csv"
queue_df.to_csv(output_path, index=False)
print(f"Successfully exported action queue to {output_path}")

# Generate and save a reusable summary figure for the paper
fig, ax = plt.subplots(figsize=(6, 4))
queue_df["archetype"].value_counts().plot(kind="bar", ax=ax, color="skyblue")
ax.set_title("Action Distribution by Content Archetype")
ax.set_xlabel("Archetype")
ax.set_ylabel("Count")
plt.tight_layout()

fig_path = "work/figures/action_archetypes.png"
plt.savefig(fig_path, dpi=300)
plt.close()
print(f"Successfully saved reuse figure to {fig_path}")
```
