## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Decision & Action: Content editors must decide which top 50 pages to refresh each week. Instead of reacting to severe traffic drops, they will prioritize work based on a decay risk score.

Cost of a Wrong Call:

False Positive (Flagging a healthy page): Wastes 4–6 hours of editor time rewriting content that didn't need updating.

False Negative (Missing a declining page): Results in lost search visibility, reduced ad/affiliate revenue, and a longer recovery cycle once rankings slip to page 2+.

Why ML/Data: Simple rule-based thresholds (e.g., "traffic dropped >20%") capture noise or seasonal shifts. An ML approach aggregates multiple tangled signals (traffic trends, click-through rates, impression decay, content freshness) to isolate genuine quality decline.
