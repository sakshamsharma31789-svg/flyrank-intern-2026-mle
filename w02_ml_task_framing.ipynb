## 4. The unit of analysis, as a real dataframe

The unit of analysis is **one SEO content item (`content_id`)**. Each row represents a single piece of published content along with its performance metrics over a 90-day window (impressions, clicks, average position, and traffic trends).

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Select the columns relevant to my lane.
lane_df = df.copy()
display(df.head().T)
numeric_columns = df.select_dtypes(include="number").columns.tolist()
print(numeric_columns)

lane_df.head()
```

Each row represents one published SEO content item (content_id). The decision is made at the document level to determine whether a page needs to be refreshed based on its historical search performance

```python

df["refresh_priority_proxy"] = (
    (df["trend_pct"] < -20) &
    (df["impressions_last_30d"] > 0)
).astype(int)

display(
    df[
        [
            "content_id",
            "trend_pct",
            "impressions_last_30d",
            "refresh_priority_proxy"
        ]
    ].head(10)
)
print(df["refresh_priority_proxy"].value_counts())
lane_df = df[
    [
        "content_id",
        "client_id",
        "main_intent",
        "content_age_days",
        "impressions_90d",
        "clicks_90d",
        "avg_position",
        "trend_pct",
        "refresh_priority_proxy"
    ]
]

display(lane_df.head(10))
```

## 5. Why ML beats a fixed rule here

A simple rule (like flagging any page down 20%) creates thousands of unranked candidates without accounting for organic search volume, competitive intent, page position, or seasonality.

Machine learning learns non-linear interactions across content age, keyword competitiveness, historical impression volume, and position drop to predict actual potential traffic recovery, allowing the editorial team to focus strictly on top-priority pages.
